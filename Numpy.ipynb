```python
import numpy as np
```

```python
a = np.array([2, 4, 6, 8, 10])
b = np.array([[2, 4, 6], [8, 10, 12]])
c = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
```

```python
print("Size:")
print(a.size)
print(b.size)
print(c.size)

print("\nNumber of dimensions:")
print(a.ndim)
print(b.ndim)
print(c.ndim)
```

```python
print("Shape of a:", a.shape)
print("After reshape:")
print(a.reshape(1, 5))
```

```python
a2 = a.copy()
print("Original:", a)
print("Copy:", a2)
```

```python
flat = b.ravel()
print("Ravel:", flat)
flat[1] = 20
print("After changing one value:", flat)
```

```python
print("Total:", a.sum())
print("Row totals:", b.sum(axis=1))
```

```python
print("Broadcasting:")
print(a + 5)
```

```python
print("Values greater than 5:")
print(a[a > 5])
```

```python
print("Positions where value is greater than 5:")
print(np.where(a > 5))
```

```python
marks = np.array([55, 72, 91, 38, 64, 84, 29, 76, 48, 95])

grade = np.select(
    [marks >= 90, marks >= 70, marks >= 50, marks >= 33],
    ["A", "B", "C", "D"],
    default="F"
)

for i in range(len(marks)):
    print("Mark:", marks[i], "Grade:", grade[i])
```
