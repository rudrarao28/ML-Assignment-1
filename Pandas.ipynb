```python
import pandas as pd
```

```python
info = {
    "Name": ["Gourav", "Kunal", "Neeraj", "Sahil", "Rudra"],
    "Age": [19, 20, 18, 21, 19],
    "City": ["Delhi", "Jaipur", "Pune", "Bhopal", "Gurugram"]
}

df = pd.DataFrame(info)
print(df)
```

```python
print("First two rows:")
print(df.head(2))

print("\nLast two rows:")
print(df.tail(2))
```

```python
print("After removing City column:")
print(df.drop(columns="City"))

print("\nAfter removing row 4:")
print(df.drop(index=4))
```

```python
print("First row:")
print(df.loc[0])
```

```python
marks = pd.DataFrame({"Marks": [72, 88, 65, 91, 54]})
df = pd.concat([df, marks], axis=1)
print(df)
```

```python
print("DataFrame description:")
print(df.describe())
```

```python
print("Null values:")
print(df.isnull().sum())
```

```python
def get_grade(mark):
    if mark >= 90:
        return "A"
    elif mark >= 75:
        return "B"
    elif mark >= 60:
        return "C"
    elif mark >= 40:
        return "D"
    else:
        return "F"

df["Grade"] = df["Marks"].apply(get_grade)
print(df)
```

```python
data2 = pd.DataFrame({
    "a": [None, 4, 7],
    "b": [2, None, 5]
})

data2 = data2.fillna(0)
print(data2)
```
